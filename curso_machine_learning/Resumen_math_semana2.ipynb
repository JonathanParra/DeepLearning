## Regresión lineal con multiples variables.

aprenderemos a hacer que la regresión lineal sea mucho más rápida y poderosa. Comencemos por analizar la versión de la regresión lineal que analiza no solo una característica, sino muchas características diferentes.

$$
f_{w,b}(x) = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b
$$

forma vectorial con notación de producto punto:

$$
f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b
$$

### Vectorization.

## Descenso del gradiente para la regresión lineal múltiple


Vamos a juntarlo todo para implementar el descenso de gradientes para la regresión lineal múltiple con vectorización

Pero en vez de pensar en J como una función de estos y otros parámetros w_j y b, escribiremos J como una función del vector de parámetros w y el número b.

$$
J(w_1, \dots, w_n, b)
$$

\mathbf{w}^T \mathbf{x}^{(i)} + b - y^{(i)}

$$

### Escalado de características


Algunas técnicas que hacen que la gran intuición funcione mucho mejor. Existe una técnica llamada escalado de características que permitirá que el descenso del gradiente sea mucho más rápido.

Veamos cómo se puede implementar el escalado de características para tomar entidades que adoptan rangos de valores muy diferentes y habilitarlas para que tengan rangos de valores comparables entre sí. ¿Cómo se escalan realmente las entidades?


Bueno, si $x_1$ oscila entre 3 y 2000

$
300 \le x_1 \le 2000
$

Dado que x_2 oscila entre 0 y 5

$
0 \le x_2 \le 5
$

una forma de obtener una versión a escala de $x_1$ es tomar cada valor original de $x_1$ y dividir por 2000, el máximo del rango. La escala $x_1$ oscilará entre 0,15 y uno.

$x_{1,\text{scaled}} = \frac{x_1}{2000}$

$
0,15 \le x_{1,\text{scaled}} \le 1
$


Por lo tanto, la escala de x_2 ahora oscilará entre 0 y 1.

$x_{2,\text{scaled}} = \frac{x_2}{5}$

$
0 \le x_{2,\text{scaled}} \le 1
$

#### Normalización Media

Además de dividir por el máximo, también puedes hacer lo que se denomina normalización de la media. Lo que se ve es que se empieza con las características originales y, a continuación, se les cambia la escala para que ambas queden centradas alrededor de cero. Mientras que antes solo tenían valores mayores que cero, ahora tienen valores negativos y positivos que suelen estar entre menos uno y más uno.

Para calcular la normalización media de $x_1$, primero encuentra la media, también llamada media de $x_1$ en tu conjunto de entrenamiento, y llamemos a esta media $\mu_1$

La formula de la Media $\mu_1$ (Promedio de la Feature $x_1$)

$
\mu_1 = \frac{1}{m}\sum_{i=1}^{m} x_1^{(i)}
$


Normalización por media usando el rango (max − min).
$$
x_1^{\text{norm}} = \frac{x_1 - \mu_1}{\max - \min}
$$

### Normalización de la Puntuación Z

Existe un último método común de reescalado llamado normalización de la puntuación Z. Para implementar la normalización de la puntuación Z, necesitas calcular algo llamado la desviación estándar de cada característica.

para implementar una normalización de la puntuación Z, primero debe calcular la media  $\mu_1$, así como la desviación estándar.

media  $\mu_1$:

$$
\mu_j = \frac{1}{m}\sum_{i=1}^{m} x_j^{(i)}
$$

Calcular la desviación estándar:

$$
\sigma_j =
\sqrt{
\frac{1}{m}
\sum_{i=1}^{m}
\left(x_j^{(i)} - \mu_j\right)^2
}
$$

Aplicar Z-score:
$$
x_j^{\text{norm}} =
\frac{x_j - \mu_j}{\sigma_j}
$$

#### ¿Por qué Z-score suele ser mejor? 

Porque:

+ No depende de valores extremos (max/min).
+ Hace que cada feature tenga $\mathrm{Varianza}(x_j^{\text{norm}}) \approx 1$
+ Mejora la condición numérica de $X^{\top} X$
+ Hace que el gradiente descendente converja más estable.

##### Intuición profunda

Después de aplicar Z-score:

Esperanza igual a 0
$$
\mathbb{E}\left[x_j^{\text{norm}}\right] = 0
$$

$$
\mathrm{Var}\left(x_j^{\text{norm}}\right) = 1
$$

Eso significa que todas las features quedan en la misma escala estadística.

## Comprobación de la convergencia del descenso de gradiente


Cuando corres un descenso en pendiente, ¿cómo puedes saber si está convergiendo? Es decir, si le ayuda a encontrar parámetros cercanos al mínimo global de la función de coste.

Si aprendemos a reconocer qué aspecto tiene una implementación de gradiente descendente que funcione bien, también podremos, elegir mejor un Alpha para una buena tasa de aprendizaje.

Como recordatorio, esta es la regla de descenso del gradiente. Una de las opciones clave es la elección de la tasa de aprendizaje ($\alpha$).

$$
\begin{aligned}
w &:= w - \alpha \frac{\partial J(w,b)}{\partial w} \\
b &:= b - \alpha \frac{\partial J(w,b)}{\partial b}
\end{aligned}
$$

Recuerda que el trabajo del gradiente descendente consiste en encontrar los parámetros w y b que, con suerte, minimicen la función de coste J. 


Lo que hago a menudo es trazar la función de coste J, que se calcula en el conjunto de entrenamiento, y grafico el valor de J en cada iteración del gradiente de descenso. Recuerde que cada iteración significa después de cada actualización simultánea de los parámetros w y b. En esta gráfica, el eje horizontal es el número de iteraciones del gradiente de descenso que ha realizado hasta ahora. Puede que obtengas una curva con este aspecto. Observe que el eje horizontal es el número de iteraciones del descenso del gradiente y no un parámetro como w o b. Esto difiere de los gráficos anteriores que ha visto, en los que el eje vertical era el costo J y el eje horizontal era un parámetro único, como w o b.

Esta curva también se denomina curva de aprendizaje. Tenga en cuenta que hay varios tipos diferentes de curvas de aprendizaje que se utilizan en el aprendizaje automático

Si el descenso del gradiente funciona correctamente, entonces el coste J debería disminuir después de cada iteración. Si J alguna vez aumenta después de una iteración, significa que Alpha está mal elegido y, por lo general, significa que Alpha es demasiado grande, o podría haber un error en el código.

### Elección del ritmo de aprendizaje

Su algoritmo de aprendizaje funcionará mucho mejor con una elección adecuada de la tasa de aprendizaje. Si es demasiado pequeño, se ejecutará muy lentamente y si es demasiado grande, es posible que ni siquiera converja.
