## Supervised Learning

Se refiere a los algoritmos que aprenden mapeos de x a y o de entrada a salida.
La caracteristica clave del aprendizaje supervisado es que le das a tu algoritmo de aprendizaje ejemplos de los que aprender.

Esto incluye las respuestas correctas, es decir la etiqueta correcta 'y' para una entrada 'x',  y es viendo los pares correctos de la entrada 'x' y la etiqueta  de salida deseada 'y', como el algoritmo de aprendizaje finalmente aprende  a tomar solo la entrada sin la etiqueta de salida y proporciona una predicción o una suposición razonablemente precisa del resultado

Tipos de Algoritmos:

+ Algoritmo de Regresion
+ Algoritmo de Clasification

### Proceso general del aprendizaje supervisado


#### Modelo de regresión lineal.
    
$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$



#### Función de costo

la función de coste nos dirá qué tan bien funciona el modelo

$$
J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right)^2
$$

función de costo usando explícitamente la hipótesis
$$
J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)^2
$$

minimizar la función de costo respecto a los parámetros 
𝑤
y 
𝑏

$$
\underset{w,b}{\operatorname{minimize}} \; J(w,b)
$$

El objetivo de la regresión lineal es encontrar los parámetros w y b que dan como resultado el menor valor posible para la función de coste J.

### Descenso del Gradiente

forma más sistemática de encontrar los valores de w y b, que diera como resultado el menor costo posible para $J(w,b)$

$$
\begin{aligned}
w &:= w - \alpha \frac{\partial J(w,b)}{\partial w} \\
b &:= b - \alpha \frac{\partial J(w,b)}{\partial b}
\end{aligned}
$$


$\alpha$ :  tasa de aprendizaje


Calculo de las derivadas:

$$
\frac{\partial J(w,b)}{\partial w}
= \frac{1}{m}\sum_{i=1}^{m}\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)x^{(i)}
$$

$$
\frac{\partial J(w,b)}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
\left(
f_{w,b}(x^{(i)}) - y^{(i)}
\right)
$$
