# Clasificación

 Resulta que la regresión lineal no es un buen algoritmo para los problemas de clasificación. Veamos por qué y esto nos llevará a un algoritmo diferente llamado regresión logística.

 Tipos de problemas de clasificación

 + **clasificación binaria** : Donde la palabra binario se refiere a que solo hay dos clases o dos categorías posibles.



### Regresión logística.

En el primer paso, espero que recuerdes que una función de línea recta, como una función de regresión lineal, se puede definir como un producto de x más b. Almacenemos este valor en una variable a la que llamaré z

$$
z = \mathbf{w} \cdot \mathbf{x} + b
$$

El siguiente paso es tomar este valor de z y pasarlo a la función sigmoidea, también llamada función logística, g.

$$
g(z) = \frac{1}{1 + e^{-z}}
$$

$$
f_{\mathbf{w},b}(\mathbf{x}) =
g(\mathbf{w} \cdot \mathbf{x} + b)
=
\frac{1}{1 + e^{-(\mathbf{w} \cdot \mathbf{x} + b)}}
$$


 Este es el modelo de regresión logística, y lo que hace es introducir una entidad o conjunto de entidades X y generar un número entre 0 y 1.

 Si algún día lees artículos de investigación o reseñas de blogs sobre todas las regresiones logísticas, a veces verás esta notación según la cual f de x es igual a p de y es igual a 1 dadas las entidades de entrada x y con los parámetros w y b. Lo que se usa aquí con el punto y coma es que w y b son parámetros que afectan al cálculo de cuál es la probabilidad de que y sea igual a 1 dada la entidad de entrada x.

 **interpretación probabilística de la regresión logística.**

$
f_{\mathbf{w},b}(\mathbf{x}) = P(y = 1 \mid \mathbf{x}; \mathbf{w}, b)
$

**Propiedad fundamental de variable binaria**

$
P(y = 0) + P(y = 1) = 1
$




## Limite de desicion (Decision Boundary)

📌 **¿Qué es el Decision Boundary?**

En regresión logística: 
$$
f_{\mathbf{w},b}(\mathbf{x}) = P(y=1 \mid \mathbf{x}; \mathbf{w}, b)
= \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}
$$

El modelo produce una probabilidad.

Pero para clasificar necesitamos una regla:
$$
\hat{y} =
\begin{cases}
1 & \text{si } P(y=1 \mid \mathbf{x}) \ge 0.5 \\
0 & \text{si } P(y=1 \mid \mathbf{x}) < 0.5
\end{cases}
$$


🎯 **¿Dónde ocurre el cambio de clase?**

Sabemos que:

$$
\sigma(z) = 0.5 \quad \text{cuando} \quad z = 0
$$

Porque:

$$
\sigma(0) = \frac{1}{1 + e^{0}} = \frac{1}{2}
$$

Entonces el punto crítico ocurre cuando:


$$
\mathbf{w}^T \mathbf{x} + b = 0
$$


📌 **Definición formal del Decision Boundary**

Esta ecuación define la frontera de decisión.

$$
\mathbf{w}^T \mathbf{x} + b = 0
$$

🔎 Interpretación geométrica

En 1 dimensión: Es un punto en la recta.

$$
wx + b = 0 \Rightarrow x = -\frac{b}{w}
$$

En 2 dimensiones: Es una línea.

$$
w_1 x_1 + w_2 x_2 + b = 0
$$

En n dimensiones: Es un hiperplano.

🚀 **Intuición clave**

La frontera no depende de la sigmoide.

Depende únicamente de:

$$
\mathbf{w}^T \mathbf{x} + b 
$$

La sigmoide solo convierte ese valor en probabilidad.

🔬 **Más profundo aún**

Si:

$
\mathbf{w}^T \mathbf{x} + b > 0 \Rightarrow y = 1
$

Si:

$
\mathbf{w}^T \mathbf{x} + b < 0 \Rightarrow y = 0
$


Entonces el modelo realmente está separando el espacio con un hiperplano lineal.


🧠 **Conexión conceptual importante**

Regresión logística es:

+ Un modelo probabilístico
+ Pero con frontera lineal

Si quieres fronteras curvas necesitas:

+ Features polinomiales
+ Kernel methods
+ Redes neuronales
